```python
import numpy as np
```

```python
# Matriks koefisien persamaan
A = np.array([
    [1, 1, 1],
    [2, 1, 1],
    [1, 2, 1]
], dtype=float)

# Nilai konstanta
B = np.array([6, 8, 9], dtype=float)

print("Matriks Koefisien:")
print(A)

print("Nilai Konstanta:")
print(B)
```

```python
# Menggabungkan matriks koefisien dan konstanta
M = np.column_stack((A, B))

n = len(B)

# Proses eliminasi Gauss
for i in range(n):

    # Menentukan pivot
    pivot = i + np.argmax(np.abs(M[i:, i]))

    # Menukar baris jika diperlukan
    M[[i, pivot]] = M[[pivot, i]]

    # Membuat elemen pivot menjadi 1
    M[i] = M[i] / M[i, i]

    # Menghilangkan elemen di bawah pivot
    for j in range(i + 1, n):
        faktor = M[j, i]
        M[j] = M[j] - faktor * M[i]

print("Hasil Eliminasi Gauss:")
print(M)
```

```python
# Menyiapkan tempat hasil
x = np.zeros(n)

# Menghitung nilai variabel dari bawah ke atas
for i in range(n - 1, -1, -1):
    x[i] = M[i, -1] - np.dot(M[i, i+1:n], x[i+1:n])

# Menampilkan hasil
print("Hasil Penyelesaian SPLTV:")
print("Nilai x =", x[0])
print("Nilai y =", x[1])
print("Nilai z =", x[2])
```
